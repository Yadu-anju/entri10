```python
import pandas as pd
df=pd.read_csv('/content/winequality-red.csv')
num_df=df[["alcohol","density"]]
num_df.head(1600)
```

Covaiance

```python
import numpy as np
covariance=np.cov(num_df.T)
print(covariance)
```

Eigon value and Eigon vecton with Covariance

```python
eigon_value,eigon_vector=np.linalg.eig(covariance)
print(eigon_value)
print(eigon_vector)
```
